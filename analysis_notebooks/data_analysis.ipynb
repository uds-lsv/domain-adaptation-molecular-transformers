```python
datasets_dir = 'pretrain_and_downstream_datasets'
out_dir = 'analysis_figures/data_analysis'
os.makedirs(out_dir, exist_ok=True)

# Load the ADME dataset and rename columns
fang_et_al = pd.read_csv(f'{datasets_dir}/ADME_public_set_3521.csv', index_col=0)
fang_ds_cols = [col for col in fang_et_al.columns if col.startswith('LOG')]

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()
for idx, ds_name in enumerate(fang_ds_cols):
    ax = axes[idx]
    ax.set_title(ds_name, fontsize=16)
    
    ds_df = fang_et_al[ds_name].dropna()
    ds_smiles = fang_et_al['SMILES'].loc[ds_df.index]
    pd.concat([ds_smiles, ds_df], axis=1).to_csv(f'fang_et_al_log_{ds_name.split(" ")[1]}.csv', index=False)
    
    # Compute common bin edges based on full dataset (with censors)
    bin_edges = np.histogram_bin_edges(ds_df, bins='auto')  # or specify bins=N

    stat_text = f"# Molecules = {len(ds_df)}"
    
    # Plot full dataset histogram
    sns.histplot(data=ds_df, bins=bin_edges, kde=True, ax=ax)
    # Drop censors and plot again (same bins!)
    censored_ds = get_censored(ds_df).dropna()
    if len(ds_df) != len(censored_ds):
        censored_smiles = fang_et_al['SMILES'].loc[censored_ds.index]
        pd.concat([censored_smiles, censored_ds], axis=1).to_csv(f'fang_et_al_log_{ds_name.split(" ")[1]}_no_censors.csv', index=False)
        sns.histplot(censored_ds, bins=bin_edges, kde=True, label='W/O Censors', ax=ax)
        stat_text += f'\n# Molecules w/o censors = {len(censored_ds)}'
        
        mean = censored_ds.mean()
        median = censored_ds.median()
        q25 = censored_ds.quantile(0.25)
        q75 = censored_ds.quantile(0.75)
    else:
        mean = ds_df.mean()
        median = ds_df.median()
        q25 = ds_df.quantile(0.25)
        q75 = ds_df.quantile(0.75)
    ax.axvline(median, color='blue', linestyle=':', label='Median')
    ax.axvline(mean, color='green', linestyle='--', label='Mean')
    ax.axvline(q25, color='black', linestyle='-', label='Q1 / Q3')
    ax.axvline(q75, color='black', linestyle='-')
    ax.legend()

    props = dict(boxstyle='round', facecolor='white', alpha=0.8, edgecolor='gray')
    ax.text(0.5, 0.95, stat_text, transform=ax.transAxes, fontsize=12,
            verticalalignment='top', horizontalalignment='center', bbox=props)
    ax.set_xlabel('')
plt.subplots_adjust(top=0.98, bottom=0.05, right=0.95, left=0.05)
plt.savefig(os.path.join(out_dir, 'fang_et_al_datasets_distributions.pdf'))
plt.show()

# store the number of molecules tested per endpoint
num_molecules = {'# Molecules': {ds_name: len(fang_et_al[ds_name].dropna()) for ds_name in fang_ds_cols}}
pd.DataFrame(num_molecules)  
```

## AsraZeneca ChEMBL
The AstraZeneca provided some challenges in properly separating the different assays of the same endpoint. We identify and separate these assays below 

```python
# Load, identify, and clean the AstraZeneca datasets
astrazeneca = pd.read_csv(f'{datasets_dir}/astrazeneca_chembl.csv', sep=';')
def detect_cl_context(desc):
    if pd.isna(desc):
        return ''
    desc = desc.lower()
    if 'microsome' in desc:
        return '_microsome'
    elif 'hepatocyte' in desc:
        return '_hepatocyte'
    else:
        return ''
organism_map = {
    pd.NA: '',  # or use np.nan if you're working with NumPy
    'Homo sapiens': 'H',
    'Canis lupus familiaris': 'D',
    'Rattus norvegicus': 'R',
    'Mus musculus': 'M',
    'Cavia porcellus': 'G'
}
# Create a new column for abbreviation codes
astrazeneca['Organism Code'] = astrazeneca['Assay Organism'].map(organism_map).fillna('')
astrazeneca['CL Context'] = astrazeneca['Assay Description'].apply(detect_cl_context)
astrazeneca['Context Label'] = astrazeneca['Organism Code'] + astrazeneca['CL Context']

astrazeneca['Prefixed Type'] = astrazeneca['Standard Type'].fillna('UNKNOWN_TYPE') + astrazeneca['Context Label']

# First, create a copy of the original to avoid modifying in place
astrazeneca_filled = astrazeneca.copy()

# Fill Standard Units using Prefixed Type logic
def infer_units(row):
    if pd.notna(row['Standard Units']):
        return row['Standard Units']
    if pd.isna(row['Prefixed Type']):
        return 'UNKNOWN_UNIT'
    return row['Prefixed Type'].split('_')[0]  # Take first part before underscore

# Apply the logic
astrazeneca_filled['Standard Units'] = astrazeneca.apply(infer_units, axis=1)

# Fill remaining missing values as usual
astrazeneca_filled = astrazeneca_filled.fillna({
    'Smiles': 'UNKNOWN_SMILES',
    'Assay Organism': 'UNKNOWN_ORGANISM',
    'Prefixed Type': 'UNKNOWN_TYPE',
    'Standard Value': -9999
})

astrazeneca_pivot = astrazeneca_filled.pivot_table(
    index=['Smiles', 'Standard Units', 'Assay Organism'],
    columns='Prefixed Type',
    values='Standard Value',
    aggfunc='mean'  # or 'first', 'max', etc.
).reset_index()
# Replace placeholders back to NaN (if needed)
astrazeneca_pivot.replace({
    'UNKNOWN_SMILES': pd.NA,
    'UNKNOWN_ORGANISM': pd.NA,
    'UNKNOWN_TYPE': pd.NA,
    -9999: pd.NA  # Only if you used this numeric placeholder
}, inplace=True)

astrazeneca_ds_cols = [col for col in astrazeneca_pivot.columns if col not in ['Smiles', 'Standard Units', 'Assay Organism']]
```

```python
fig, axes = plt.subplots(4, 4, figsize=(20, 20))
axes = axes.flatten()
for idx, ds_name in enumerate(astrazeneca_ds_cols):
    ds_name_save = ds_name
    ax = axes[idx]
    ax.set_title(ds_name, fontsize=16)
    ds_df = astrazeneca_pivot[['Smiles', 'Standard Units', ds_name]].dropna().copy()
    measure_unit = ds_df['Standard Units'].unique()
    if len(measure_unit) == 1: 
        measure_unit = measure_unit[0]
    else:
        print(ds_name, measure_unit)
    ax.set_xlabel(measure_unit, fontsize=14)
    
    unlogged = False
    if 'Log' not in ds_name:
        try:
            logged_vals = np.log10(ds_df[ds_name])
            ds_df[ds_name] = logged_vals
            ax.set_xlabel(f'Log({measure_unit})', fontsize=14)
            ds_name_save = f'log_{ds_name_save}'
            #ds_df[['Smiles', ds_name]].to_csv(f'astrazeneca_log_{ds_name}.csv')
        except Exception as e:
            print(f"[DEBUG] Skipping log10 for {ds_name} due to {type(e).__name__}: {e}")
            unlogged = True
            #ds_df[['Smiles', ds_name]].to_csv(f'astrazeneca_{ds_name}.csv')
    
    ds_df[['Smiles', ds_name]].to_csv(f'astrazeneca_{ds_name_save}.csv', index=False)
    ds_values = ds_df[ds_name]
    
    # Compute common bin edges based on full dataset (with censors)
    bin_edges = np.histogram_bin_edges(ds_values, bins='auto')  # or specify bins=N

    stat_text = f"# Molecules = {len(ds_values)}"
    
    # Plot full dataset histogram
    sns.histplot(data=ds_values, bins=bin_edges, kde=True, ax=ax)

    # Drop censors and plot again (same bins!)
    censored_ds = get_censored(ds_values, censor_percent=5).dropna()
    if len(ds_values) != len(censored_ds):
        uncensored_smiles = ds_df['Smiles'].loc[censored_ds.index]
        pd.concat([uncensored_smiles, censored_ds], axis=1).to_csv(f'astrazeneca_{ds_name_save}_no_censors.csv', index=False)
        sns.histplot(censored_ds, bins=bin_edges, kde=True, label='W/O Censors', ax=ax)
        stat_text += f'\n# Molecules w/o censors = {len(censored_ds)}'
        mean = censored_ds.mean()
        median = censored_ds.median()
        q25 = censored_ds.quantile(0.25)
        q75 = censored_ds.quantile(0.75)
    else:
        mean = ds_values.mean()
        median = ds_values.median()
        q25 = ds_values.quantile(0.25)
        q75 = ds_values.quantile(0.75)
    ax.axvline(median, color='blue', linestyle=':', label='Median')
    ax.axvline(mean, color='green', linestyle='--', label='Mean')
    ax.axvline(q25, color='black', linestyle='-', label='Q1 / Q3')
    ax.axvline(q75, color='black', linestyle='-')
    ax.legend()
    ax.legend()
    
    props = dict(boxstyle='round', facecolor='white', alpha=0.8, edgecolor='gray')
    ax.text(0.5, 0.95, stat_text, transform=ax.transAxes, fontsize=12,
                    verticalalignment='top', horizontalalignment='center', bbox=props)
plt.subplots_adjust(top=0.98, bottom=0.05, right=0.95, left=0.05)
plt.savefig(os.path.join(out_dir, 'astrazeneca_datasets_distributions.pdf'))
plt.show()

# store the number of molecules tested per endpoint
num_molecules = {'# Molecules': {ds_name: len(astrazeneca_pivot[ds_name].dropna()) for ds_name in astrazeneca_ds_cols}}
pd.DataFrame(num_molecules)  
```

# Final selection  
Below is the final selection of the two data sources (# molecules > 1500)

```python
# --- Define renaming dictionaries ---
fang_ds_cols = {
    "LOG SOLUBILITY PH 6.8 (ug/mL)": 'Solubility$_{Fang\_et\_al}$',
    "LOG MDR1-MDCK ER (B-A/A-B)": 'Permeability',
    "LOG HLM_CLint (mL/min/kg)": 'HLM',
    "LOG RLM_CLint (mL/min/kg)": 'RLM',
}

fang_target_value = {
    'Solubility$_{Fang\_et\_al}$': 'Log($\mu$g/mL)',
    'Permeability': 'Log(efflux ratio)',
    'HLM': 'Log(mL/min/kg)',
    'RLM': 'Log(mL/min/kg)',
}

astrazeneca_ds_cols = {
    'PPBH': 'HPPB',
    'Solubility': 'Solubility$_{Astrazeneca\\_ChEMBL}$',
    'LogD7.4': 'Lipophilicity'
}

astrazeneca_target_values = {
    'HPPB': 'Log(% Bound)',
    'Solubility$_{Astrazeneca\\_ChEMBL}$': 'Log(ng/mL)',
    'Lipophilicity': 'LogD7.4'
}

# --- Step 1: Extract SMILES + relevant data from each source ---
fang_df = fang_et_al[list(fang_ds_cols.keys()) + ['SMILES']].copy()
az_df = astrazeneca_pivot[list(astrazeneca_ds_cols.keys()) + ['Smiles']].copy()

# --- Step 2: Log-transform AstraZeneca cols not starting with 'Log' ---
cols_to_log = [col for col in az_df.columns if col in astrazeneca_ds_cols and not col.startswith('Log')]
az_df[cols_to_log] = az_df[cols_to_log].apply(np.log10)

# --- Step 3: Rename columns including SMILES ---
renaming_dict = {**fang_ds_cols, **astrazeneca_ds_cols, 'Smiles': 'SMILES'}
fang_df.rename(columns=renaming_dict, inplace=True)  # identity rename, just to be consistent
az_df.rename(columns=renaming_dict, inplace=True)

# --- Step 4: Combine both into one DataFrame ---
experiment_df = pd.concat([fang_df, az_df], ignore_index=True)

experiment_df.head()
```

```python
target_values = fang_target_value | astrazeneca_target_values
datasets = ['Lipophilicity', 'Permeability', 'Solubility$_{Fang\\_et\\_al}$', 'Solubility$_{Astrazeneca\\_ChEMBL}$', 'HLM', 'RLM', 'HPPB']
```

```python
fig, axes = plt.subplots(3, 3, figsize=(16, 12))
axes = axes.flatten()

legend_dict = {}  # label -> handle

for idx, dataset in enumerate(datasets):
    ax = axes[idx]
    ax.set_title(dataset)
    
    ds_values = experiment_df[dataset].dropna()
    
    bin_edges = np.histogram_bin_edges(ds_values, bins='auto')
    stat_text = f"# Molecules = {len(ds_values)}"
    
    # Plot all data
    sns.histplot(data=ds_values, bins=bin_edges, kde=True, ax=ax, label='All Molecules')
    
    # Replace lines with boxplot on a secondary x-axis
    ax_box = ax.twinx()
    ax_box.set_xlim(ax.get_xlim())
    # Set y-limits so we can control where the boxplot appears
    ax_box.set_ylim(0, 1)  # Important!
    y_pos = 0.2
    ax_box.boxplot(ds_values, vert=False, positions=[y_pos], widths=0.1,
            patch_artist=True,
            boxprops=dict(facecolor='none', alpha=0.7, color='blue'),
            medianprops=dict(color='black', linewidth=1))
    ax_box.plot(np.mean(ds_values), y_pos, '^', color='green', markersize=5, label='Mean')
    ax_box.set(yticks=[], ylabel='')

    # Censor handling
    censored_ds = get_censored(ds_values, censor_percent=5).dropna()
    if len(ds_values) != len(censored_ds):
        sns.histplot(censored_ds, bins=bin_edges, kde=True, ax=ax, label='Molecules without censored labels')
        y_pos = 0.4
        ax_box.boxplot(censored_ds, vert=False, positions=[y_pos], widths=0.1,
            patch_artist=True,
            boxprops=dict(facecolor='none', alpha=0.7, color='brown'),
            medianprops=dict(color='black', linewidth=1))
        ax_box.plot(np.mean(censored_ds), y_pos, '^', color='green', markersize=5, label='Mean')
        
        stat_text += f'\n# Molecules w/o censors = {len(censored_ds)}'
        values = censored_ds
    else:
        values = ds_values

    
    # Text box
    props = dict(boxstyle='round', facecolor='white', alpha=0.8, edgecolor='gray')
    ax.text(0.5, 0.95, stat_text, transform=ax.transAxes, fontsize=12,
            verticalalignment='top', horizontalalignment='center', bbox=props)

    ax.set_xlabel(target_values[dataset])

    # --- Collect all legend handles/labels globally ---
    handles, labels = ax.get_legend_handles_labels()
    for h, l in zip(handles, labels):
        legend_dict[l] = h  # update or add

# --- Hide unused subplots ---
for i in range(len(datasets), len(axes)):
    axes[i].axis('off')

# --- Create ordered legend from collected entries ---
legend_order = ['All molecules', 'Molecules without censored labels']
ordered_labels = [l for l in legend_order if l in legend_dict] + [
    l for l in legend_dict if l not in legend_order
]
ordered_handles = [legend_dict[l] for l in ordered_labels]

# --- Add global legend ---
fig.legend(
    handles=ordered_handles,
    labels=ordered_labels,
    loc='upper center',
    bbox_to_anchor=(0.6, 0.25),
    ncol=3,
    fontsize=16,
    frameon=False
)

# --- Layout & Save ---
plt.subplots_adjust(hspace=0.35, wspace=0.3)
plt.savefig(os.path.join(out_dir, 'endpoints_summary.pdf'), bbox_inches='tight')
plt.show()

```

## Quantifying both datasets

Here, we show the number of molecules and unique Murcko scaffold per dataset to examine each dataset diversity. We also show the intersection between molecules and scaffolds to examine the diversity between the two datasets. 
* NB: We did not perform our own standardization on the molecules from GuacaMol and ADME. So, there is a risk that the same molecule exists in both datasets, but undetectable by the intersection analysis 
* NB: extracting scaffolds by RDKit requires standardization first, so, all shared scaffolds should be detected correctly. However, it is not guaranteed that the model has detected these scaffolds (since the molecules were not standardized befor given to the model)


```python
counts = {
    'GuacaMol': {'# Molecules': len(guacamol), 
                 '# Unique Murcko scaffolds': len(guacamol['Scaffold'].unique())},
    }
for dataset in ['Lipophilicity', 'Solubility$_{Astrazeneca\\_ChEMBL}$', 'HPPB', 'Permeability', 'Solubility$_{Fang\\_et\\_al}$', 'HLM', 'RLM']:
    ds_df = adme.dropna(subset=[dataset])
    len_df = len(ds_df)
    len_unique_scaffolds = len(ds_df['Scaffold'].unique())
    pct_unique_scaffolds = (len_unique_scaffolds / len_df) * 100
    guacamol_intersect_ds_mols =  len(set(guacamol['SMILES']).intersection(set(ds_df['SMILES'])))
    pct_common_mols = (guacamol_intersect_ds_mols / len_df) * 100
    guacamol_intersect_ds_scaffolds = len(set(guacamol['Scaffold']).intersection(set(ds_df['Scaffold'])))
    pct_common_scaffolds = (guacamol_intersect_ds_scaffolds / len_unique_scaffolds) * 100
    counts[dataset] = {
        '# Molecules': len_df,
        '# Unique Murcko scaffolds': len_unique_scaffolds,
        '% Unique Scaffolds': pct_unique_scaffolds,
        '# Common Molecules': guacamol_intersect_ds_mols,
        '% Common molecules': pct_common_mols,
        '# Common scaffolds': guacamol_intersect_ds_scaffolds,
        '% Common scaffolds': pct_common_scaffolds}
pd.DataFrame(counts).T.round(2)
```
